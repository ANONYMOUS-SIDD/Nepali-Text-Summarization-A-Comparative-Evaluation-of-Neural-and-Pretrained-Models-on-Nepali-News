```python
# mT5-small for Nepali Abstractive Summarization
#
# Dependencies (install once):
#   pip install torch pandas numpy scikit-learn transformers \
#               datasets sentencepiece accelerate \
#               sacrebleu bert-score tqdm
#
# Environment:
#   NEPALI_DATA_DIR   folder containing small.csv, medium.csv, large.csv
#                     (default: ./data/)
#   NEPALI_WORK_DIR   output folder (default: ./mt5_workspace/)
#
# T5 special tokens:
#   decoder_start_token_id = pad_token_id
#   eos_token_id           = tokenizer.eos_token_id
#   pad_token_id           = tokenizer.pad_token_id
#
# Summary tokenization uses `text_target=` (T5 convention).
# Decoding: beam search with beam size 4.

import os
import random
from pathlib import Path
from collections import Counter

import numpy as np
import pandas as pd
import torch
from torch.utils.data import Dataset, DataLoader
from sklearn.model_selection import train_test_split
from datasets import Dataset as HFDataset
from transformers import (
    AutoTokenizer,
    AutoModelForSeq2SeqLM,
    DataCollatorForSeq2Seq,
    Seq2SeqTrainer,
    Seq2SeqTrainingArguments,
)
from sacrebleu.metrics import BLEU
from bert_score import score as bert_score_fn
from tqdm.auto import tqdm


# Configuration
CONFIG = {
    "data_dir":   Path(os.environ.get("NEPALI_DATA_DIR", "./data")),
    "work_dir":   Path(os.environ.get("NEPALI_WORK_DIR", "./mt5_workspace")),
    "base_model": "google/mt5-small",

    # 80:10:10 split
    "test_size":      0.2,
    "seed":           42,
    "max_input_len":  512,
    "max_target_len": 128,

    # Common fine-tuning configuration for all pretrained models
    "train_batch_size":    8,
    "eval_batch_size":     8,
    "grad_accum":          1,
    "epochs":              3,
    "lr":                  5e-5,
    "fp16":                True,
    "logging_steps":       500,
    "generation_max_len":  128,

    # Decoding
    "infer_max_len":    128,
    "infer_num_beams":  4,

    # Inference batching
    "infer_batch_size":  8,
    "infer_num_workers": 2,
}


# Reproducibility
random.seed(CONFIG["seed"])
np.random.seed(CONFIG["seed"])
torch.manual_seed(CONFIG["seed"])
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(CONFIG["seed"])
torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
if device.type != "cuda":
    raise RuntimeError("CUDA GPU required.")

WORK_DIR        = CONFIG["work_dir"]
WORK_DIR.mkdir(parents=True, exist_ok=True)
CKPT_DIR        = WORK_DIR / "mt5-checkpoints"
FINAL_MODEL_DIR = WORK_DIR / "mt5-final"
TEST_RAW_CSV    = WORK_DIR / "mt5_test_raw.csv"
PRED_CSV        = WORK_DIR / "mt5_test_predictions.csv"
SCORES_CSV      = WORK_DIR / "mt5_final_scores.csv"


# Data: load, clean, split
def load_and_split(cfg):
    data_dir = cfg["data_dir"]
    required = ("small.csv", "medium.csv", "large.csv")
    missing = [n for n in required if not (data_dir / n).is_file()]
    if missing:
        raise FileNotFoundError(
            f"Missing {missing} in {data_dir.resolve()}. "
            f"Set NEPALI_DATA_DIR or place the CSVs in ./data/."
        )

    df = pd.concat([pd.read_csv(data_dir / n) for n in required], ignore_index=True)
    df = df.dropna(subset=["content", "summary"]).copy()
    df["content"] = df["content"].astype(str).str.strip()
    df["summary"] = df["summary"].astype(str).str.strip()
    df = df[(df["content"] != "") & (df["summary"] != "")]
    df = df.drop_duplicates(subset=["content"], keep="first").reset_index(drop=True)

    train_df, temp_df = train_test_split(
        df, test_size=cfg["test_size"], random_state=cfg["seed"]
    )
    val_df, test_df = train_test_split(
        temp_df, test_size=0.5, random_state=cfg["seed"]
    )
    return train_df, val_df, test_df


train_df, val_df, test_df = load_and_split(CONFIG)
test_df[["content", "summary"]].to_csv(TEST_RAW_CSV, index=False, encoding="utf-8")


# Tokenizer
tokenizer = AutoTokenizer.from_pretrained(CONFIG["base_model"])


# Load or train model
if (FINAL_MODEL_DIR / "config.json").is_file():
    model = AutoModelForSeq2SeqLM.from_pretrained(FINAL_MODEL_DIR).to(device)
    tokenizer = AutoTokenizer.from_pretrained(FINAL_MODEL_DIR)
else:
    model = AutoModelForSeq2SeqLM.from_pretrained(CONFIG["base_model"])
    model.gradient_checkpointing_enable()

    def tokenize_fn(examples):
        inputs = tokenizer(
            examples["content"],
            max_length=CONFIG["max_input_len"],
            truncation=True,
            padding=False,
        )
        labels = tokenizer(
            text_target=examples["summary"],
            max_length=CONFIG["max_target_len"],
            truncation=True,
            padding=False,
        )
        inputs["labels"] = labels["input_ids"]
        return inputs

    train_ds = HFDataset.from_pandas(train_df[["content", "summary"]])
    val_ds   = HFDataset.from_pandas(val_df[["content", "summary"]])
    tokenized_train = train_ds.map(tokenize_fn, batched=True, remove_columns=["content", "summary"])
    tokenized_val   = val_ds.map(tokenize_fn, batched=True, remove_columns=["content", "summary"])

    data_collator = DataCollatorForSeq2Seq(
        tokenizer=tokenizer, model=model, padding=True,
    )

    training_args = Seq2SeqTrainingArguments(
        output_dir=str(CKPT_DIR),
        eval_strategy="epoch",
        save_strategy="epoch",
        logging_strategy="steps",
        logging_steps=CONFIG["logging_steps"],
        per_device_train_batch_size=CONFIG["train_batch_size"],
        per_device_eval_batch_size=CONFIG["eval_batch_size"],
        gradient_accumulation_steps=CONFIG["grad_accum"],
        num_train_epochs=CONFIG["epochs"],
        learning_rate=CONFIG["lr"],
        fp16=CONFIG["fp16"],
        predict_with_generate=True,
        generation_max_length=CONFIG["generation_max_len"],
        load_best_model_at_end=True,
        metric_for_best_model="eval_loss",
        greater_is_better=False,
        report_to="none",
    )

    trainer = Seq2SeqTrainer(
        model=model,
        args=training_args,
        train_dataset=tokenized_train,
        eval_dataset=tokenized_val,
        data_collator=data_collator,
    )
    try:
        trainer.tokenizer = tokenizer
    except AttributeError:
        pass

    trainer.train()

    model.save_pretrained(FINAL_MODEL_DIR)
    tokenizer.save_pretrained(FINAL_MODEL_DIR)

model = model.to(device).eval()
# T5 uses pad as decoder start.
model.config.decoder_start_token_id = tokenizer.pad_token_id
model.config.eos_token_id           = tokenizer.eos_token_id
model.config.pad_token_id           = tokenizer.pad_token_id


# Batch inference
class ContentDataset(Dataset):
    def __init__(self, frame):
        self.texts = frame["content"].astype(str).tolist()

    def __len__(self):
        return len(self.texts)

    def __getitem__(self, i):
        return self.texts[i]


def make_collate(tok, cfg):
    def collate(batch):
        return tok(
            batch,
            max_length=cfg["max_input_len"],
            truncation=True,
            padding=True,
            return_tensors="pt",
        )
    return collate


test_loader = DataLoader(
    ContentDataset(test_df),
    batch_size=CONFIG["infer_batch_size"],
    shuffle=False,
    collate_fn=make_collate(tokenizer, CONFIG),
    num_workers=CONFIG["infer_num_workers"],
    pin_memory=True,
)

# Beam search with beam size 4.
gen_kwargs = dict(
    max_length=CONFIG["infer_max_len"],
    num_beams=CONFIG["infer_num_beams"],
    early_stopping=True,
    pad_token_id=tokenizer.pad_token_id,
    eos_token_id=tokenizer.eos_token_id,
    decoder_start_token_id=tokenizer.pad_token_id,
)

predictions = []
for batch in tqdm(test_loader, desc="Generating mT5"):
    with torch.no_grad():
        gen_ids = model.generate(
            batch["input_ids"].to(device),
            attention_mask=batch["attention_mask"].to(device),
            **gen_kwargs,
        )
    predictions.extend(tokenizer.batch_decode(gen_ids, skip_special_tokens=True))

pred_df = pd.DataFrame({
    "source":    test_df["content"].astype(str),
    "reference": test_df["summary"].astype(str),
    "predicted": predictions,
})
pred_df.to_csv(PRED_CSV, index=False, encoding="utf-8")


# Evaluation
pred_df = pd.read_csv(PRED_CSV).fillna("")
empty = pred_df["predicted"].astype(str).str.strip() == ""
n_empty = int(empty.sum())
if n_empty:
    pred_df.loc[empty, "predicted"] = "।"

refs = pred_df["reference"].astype(str).tolist()
hyps = pred_df["predicted"].astype(str).tolist()

bleu_f  = BLEU(tokenize="flores200", max_ngram_order=4).corpus_score(hyps, [refs]).score
bleu_13 = BLEU(tokenize="13a",      max_ngram_order=4).corpus_score(hyps, [refs]).score


# ROUGE (manual F1)
def _ngrams(tokens, n):
    return [tuple(tokens[i:i + n]) for i in range(len(tokens) - n + 1)]


def rouge_n_f1(hyp, ref, n):
    h_ng, r_ng = _ngrams(hyp, n), _ngrams(ref, n)
    if not h_ng or not r_ng:
        return 0.0
    overlap = sum((Counter(h_ng) & Counter(r_ng)).values())
    p, r = overlap / len(h_ng), overlap / len(r_ng)
    return 0.0 if p + r == 0 else 2 * p * r / (p + r)


def rouge_l_f1(hyp, ref):
    m, n = len(hyp), len(ref)
    if m == 0 or n == 0:
        return 0.0
    dp = [[0] * (n + 1) for _ in range(m + 1)]
    for i in range(1, m + 1):
        for j in range(1, n + 1):
            dp[i][j] = (
                dp[i - 1][j - 1] + 1 if hyp[i - 1] == ref[j - 1]
                else max(dp[i - 1][j], dp[i][j - 1])
            )
    lcs = dp[m][n]
    p, r = lcs / m, lcs / n
    return 0.0 if p + r == 0 else 2 * p * r / (p + r)


pairs = [(h.split(), r.split()) for h, r in zip(hyps, refs)]
r1 = float(np.mean([rouge_n_f1(h, r, 1) for h, r in pairs]))
r2 = float(np.mean([rouge_n_f1(h, r, 2) for h, r in pairs]))
rL = float(np.mean([rouge_l_f1(h, r)    for h, r in pairs]))


# BERTScore using bert-base-multilingual-cased.
P, R, F1 = bert_score_fn(
    hyps, refs,
    model_type="bert-base-multilingual-cased",
    verbose=False,
    device=device.type,
)


# Jaccard
def _jaccard(A, B):
    if not A and not B:
        return 1.0
    if not A or not B:
        return 0.0
    return len(A & B) / len(A | B)


def jaccard_word(a, b):
    return _jaccard(set(str(a).split()), set(str(b).split()))


def jaccard_char_bigram(a, b):
    a, b = str(a), str(b)
    A = {a[i:i + 2] for i in range(max(len(a) - 1, 0))}
    B = {b[i:i + 2] for i in range(max(len(b) - 1, 0))}
    return _jaccard(A, B)


jac_w = float(np.mean([jaccard_word(h, r)        for h, r in zip(hyps, refs)]))
jac_c = float(np.mean([jaccard_char_bigram(h, r) for h, r in zip(hyps, refs)]))


# Save results
summary = {
    "model":               "mT5-small (fine-tuned)",
    "sacrebleu_flores200": round(bleu_f,  4),
    "sacrebleu_13a":       round(bleu_13, 4),
    "rouge1":              round(r1, 4),
    "rouge2":              round(r2, 4),
    "rougeL":              round(rL, 4),
    "bertscore_p":         round(P.mean().item(),  4),
    "bertscore_r":         round(R.mean().item(),  4),
    "bertscore_f1":        round(F1.mean().item(), 4),
    "jaccard_word":        round(jac_w, 4),
    "jaccard_char":        round(jac_c, 4),
    "n_samples":           len(refs),
    "n_empty":             n_empty,
    "avg_pred_chars":      round(float(np.mean([len(h) for h in hyps])), 1),
    "avg_ref_chars":       round(float(np.mean([len(r) for r in refs])), 1),
}

pd.Series(summary).to_csv(SCORES_CSV)
```
